## **Connecting to MySQL using Python**

```python
# importing libraries

import mysql.connector
from mysql.connector import Error
import pandas as pd
```

```python
def create_server_connection(hostname, username, password):
    connection = None
    try:
        connection = mysql.connector.connect(
            host = hostname,
            user = username,
            passwd = password
        )
        print("MySQL Database connection successful.")
    except Error as err:
        print(f"Error: {err}")

    return connection

# assign MySQL credentials
pw = "root"
# Database name
db = "mysql_python"

connection = create_server_connection("localhost", "root", pw)
```

```python
# Create mysql_python database now

def create_database(connection, query):
    cursor = connection.cursor()
    try:
        cursor.execute(query)
        print("Database created successfully.")
    except Error as err:
        print(f"Error: {err}")

create_database_query = "Create database mysql_python"
create_database(connection, create_database_query)
```

```python
# Connect to database now

def create_db_connection(hostname, username, password, db_name):
    connection = None
    try:
        connection = mysql.connector.connect(
            host = hostname,
            user = username,
            passwd = password,
            database = db_name
        )
        print("MySQL database connection successful")
    except Error as err:
        print(f"Error: {err}")

    return connection

create_db_connection("localhost", "root", pw, db)
```

```python
# Executiong SQL queries now

def execute_query(connection, query):
    cursor = connection.cursor()
    try:
        cursor.execute(query)
        connection.commit()
        print("Query was successful")
    except Error as err:
        print(f"Error: {err}")
```

#### **Actual queries from here onwards**

```python
create_orders_table = '''
    create table orders(
    orderid int primary key,
    customer_name varchar(30) not null,
    product_name varchar(20) not null,
    date_ordered Date,
    quantity int,
    unit_price float,
    phone_number varchar(20)
    );
'''

# connect to database first
connection = create_db_connection("localhost", "root", pw, db)
execute_query(connection, create_orders_table)
```

```python
# Inserting records

data_orders = '''
    insert into orders values
    (101, 'Steve', 'Bag', '2024-12-12', 3, 2300, '987654321'),
    (102, 'Stacy', 'Handbag', '2025-01-12', 2, 4500, '987564321'),
    (103, 'David', 'Smart TV', '2025-01-01', 1, 5000, '978654321'),
    (104, 'Thomas', 'Headphones', '2024-10-17', 1, 2000, '897654321'),
    (105, 'Bobby', 'T-Shirts', '2025-02-14', 5, 3000, '999654321'),
    (106, 'Angela', 'Microwave Oven', '2025-01-14', 1, 3250, '919654321');
'''

connection = create_db_connection('localhost', 'root', pw, db)
execute_query(connection, data_orders)
```

```python
# Reading queries

def read_query(connection, query):
    cursor = connection.cursor()
    result = None
    try:
        cursor.execute(query)
        result = cursor.fetchall()
        return result
    except Error as err:
        print(f"Error: {err}")
```

**Using the select statement**

```python
q1 = '''
    select * from orders;
'''

connection = create_db_connection('localhost', 'root', pw, db)
results = read_query(connection, q1)


for result in results:
    print(result)
```

**Individual columns from the table**

```python
q2 = '''
    select customer_name, phone_number from orders
'''

results = read_query(connection, q2)


for result in results:
    print(result)
```

**year from the data ordered column**

```python
q3 = '''
    select year(date_ordered) from orders
'''

results = read_query(connection, q3)


for result in results:
    print(result)
```

```python
q4 = '''
    select distinct year(date_ordered) from orders
'''

results = read_query(connection, q4)


for result in results:
    print(result)
```

```python
q5 = '''
    select * from orders
    where year(date_ordered) < '2025';   
'''

# all orders before 2025

results = read_query(connection, q5)


for result in results:
    print(result)
```

```python
q6 = '''
    select * from orders
    where year(date_ordered) >= '2025';   
'''

# all orders after 2025

results = read_query(connection, q6)


for result in results:
    print(result)
```

```python
q7 = '''
    select * from orders
    order by unit_price desc;   
'''

# order by unit price

results = read_query(connection, q7)


for result in results:
    print(result)
```

```python
# Creating dataframes

from_db = []

for result in results:
    result = list(result)
    from_db.append(result)

columns = ["order_id", "customer_name", "product_name", "date_ordered", "quantity",
          "unit_price", "phone_number"]

df = pd.DataFrame(from_db, columns=columns)
display(df)
```

```python
# update command

# change unit price of Mircrowave from 3250 to 3000

update = '''
    update orders
    set unit_price = 3000
    where orderid = 106;
'''

execute_query(connection, update)

```

```python
# checking for the changes
q8 = '''
    select * from orders;
'''

connection = create_db_connection('localhost', 'root', pw, db)
results = read_query(connection, q8)


for result in results:
    print(result)
```
